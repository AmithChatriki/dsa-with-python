```python
import ctypes
```

```python
class MyList:

    def __init__(self):
        self.size = 1
        self.n = 0

        self.A = self.__create_array(self.size)
    
    def __create_array(self, capacity):
        # Creates a C type array (static, referential) with size capacity
        return (capacity*ctypes.py_object)()

    def __len__(self):
        return self.n
    
    def append(self, item):
        if self.n == self.size:
            # resize
            self.__resize(self.size*2)
        # append
        self.A[self.n] = item
        self.n += 1
    
    def pop(self):
        # Delete the last the element
        if self.n == 0:
            return 'List is Empty'
        
        print(self.A[self.n-1])
        self.n = self.n-1

    def clear(self):
        self.n = 0
        self.size = 1
    
    def find(self, item):
        for i in range(self.n):
            if self.A[i] == item:
                return i
        return 'ValueError: Item not found in list'
    
    def insert(self, index, item):
        if self.n == self.size:
            self.__resize(self.size*2)

        for i in range(self.n, index, -1):
            self.A[i] = self.A[i-1]
        
        self.A[index] = item
        self.n += 1
    
    def remove(self, item):
        # Delete an item
        pos = self.find(item)
        if type(pos) == int:
            self.__delitem__(pos)
        else:
            return pos
    
    def __delitem__(self, pos):
        # Delete an item at the index
        if 0 <= pos < self.n:
            for i in range(pos, self.n-1):
                self.A[i] = self.A[i+1]
            
            self.n = self.n-1

    def __resize(self, new_capacity):
        # Create a new array with the new_capacity
        B = self.__create_array(new_capacity)
        self.size = new_capacity

        # Copy content from A to B
        for i in range(self.n):
            B[i] = self.A[i]
        
        # Re-assign
        self.A = B
    
    def __str__(self):
        # [1, 2, 3]
        result = ""
        for i in range(self.n):
            result = result + str(self.A[i]) + ", "
        return '[' + result[:-2] + ']'

    def __getitem__(self, index):
        # Indexing
        if 0 <= index < self.n:
            return self.A[index]
        else:
            return 'IndexError: Index out of range'
```
